```python
catalog_name = "databrickstask"
path = f"/Volumes/{catalog_name}/raw_files/sourse_data"
df_taxi = spark.read.parquet(f"{path}/yellow_tripdata_2024-01.parquet")
df_taxi.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable(f"{catalog_name}.bronze.taxi")

df_users = spark.read.format("avro").load(f"{path}/users.avro")
df_users.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable(f"{catalog_name}.bronze.users")

```
