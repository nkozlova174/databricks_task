```python
df_silver_taxi = spark.table("databrickstask.silver.taxi")
df_silver_drivers = spark.table("databrickstask.silver.drivers")
df_silver_drivers_for_join = (
    df_silver_drivers
    .withColumnRenamed("ingest_time", "driver_ingest_time") 
    .withColumnRenamed("name", "driver_name")
    .withColumnRenamed("car_number", "driver_car_number")
    .withColumnRenamed("experience", "driver_experience")
    .withColumnRenamed("timestamp", "driver_timestamp"))

df_enriched_trips = df_silver_taxi.join(df_silver_drivers_for_join, on="driver_id", how="left")

(
    df_enriched_trips.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("databrickstask.gold.enriched_trips")
)
```
